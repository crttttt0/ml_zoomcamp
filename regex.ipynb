## Задача 1. Числа в тексте

Верни список всех чисел из текста **в виде строк**. Числа бывают целые и дробные (через точку), могут быть отрицательными.

Точка в конце предложения частью числа не считается: `'Их 10.'` → `['10']`.

**Тема:** `\d`, квантификаторы, необязательная часть `?`, `re.findall`.

````python
def find_numbers(text: str) -> list[str]:
    res = re.findall(r"(?:-*\d+)(?:\.+\d+)*", text)
    print(res)

    return res
````

````python
assert find_numbers("Температура -5.5, вчера 12, а завтра 3.14") == [
    "-5.5",
    "12",
    "3.14",
]
assert find_numbers("Их 10. Конец") == ["10"]
assert find_numbers("цифр нет") == []
assert find_numbers("x=-7;y=0.25;z=100") == ["-7", "0.25", "100"]
print("OK")
````
